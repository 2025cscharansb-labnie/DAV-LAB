```python
import pandas as pd

# Assuming the uploaded file is 'research_student (1) (1).xlsx'
file_name = 'research_student (1) (1).xlsx'
df = pd.read_excel(file_name)

display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Drop rows where 'Marks[10th]' is NaN to ensure the histogram plots correctly
df_marks_10th = df.dropna(subset=['Marks[10th]'])

plt.figure(figsize=(10, 6))
sns.histplot(df_marks_10th['Marks[10th]'], kde=True, bins=15)
plt.title('Distribution of 10th Grade Marks')
plt.xlabel('Marks[10th]')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

This histogram shows the distribution of 10th-grade marks. The `kde=True` argument adds a Kernel Density Estimate line, which provides a smoothed representation of the distribution. From this, you can see if the marks are normally distributed, skewed, or if there are multiple peaks.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['Marks[10th]'].dropna())
plt.title('Box Plot of 10th Grade Marks')
plt.ylabel('Marks[10th]')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['Marks[12th]'].dropna())
plt.title('Box Plot of 12th Grade Marks')
plt.ylabel('Marks[12th]')
plt.show()
```

This box plot illustrates the distribution of 12th-grade marks, similar to how we visualized 10th-grade marks. You can compare the spread, median, and potential outliers between the two.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['GPA 1'].dropna())
plt.title('Box Plot of GPA 1')
plt.ylabel('GPA 1')
plt.show()
```

Here's a box plot for 'GPA 1'. This helps you understand the central tendency, variability, and any potential extreme values in the first-semester GPA of the students.

```python
plt.figure(figsize=(12, 7))
sns.boxplot(x='Branch', y='GPA 1', data=df.dropna(subset=['Branch', 'GPA 1']), palette='viridis')
plt.title('GPA 1 Distribution Across Branches')
plt.xlabel('Branch')
plt.ylabel('GPA 1')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Drop rows where 'Gender' is NaN to ensure the countplot plots correctly
df_gender = df.dropna(subset=['Gender'])

plt.figure(figsize=(8, 6))
sns.countplot(x='Gender', data=df_gender, palette='viridis')
plt.title('Count of Students by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

This count plot illustrates the number of students for each gender category in your dataset. It's useful for quickly understanding the demographic composition of your student data.

```python
plt.figure(figsize=(10, 7))
sns.scatterplot(x='Marks[10th]', y='Marks[12th]', data=df.dropna(subset=['Marks[10th]', 'Marks[12th]']))
plt.title('10th Grade Marks vs. 12th Grade Marks')
plt.xlabel('Marks[10th]')
plt.ylabel('Marks[12th]')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

This scatter plot helps visualize the relationship between a student's 10th-grade marks and their 12th-grade marks. A clear upward trend would suggest a positive correlation between performance in these two stages.

```python
# Drop rows where 'Category' is NaN to ensure the countplot plots correctly
df_category = df.dropna(subset=['Category'])

plt.figure(figsize=(10, 6))
sns.countplot(x='Category', data=df_category, palette='magma', order=df_category['Category'].value_counts().index)
plt.title('Count of Students by Category')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(16, 12))
sns.heatmap(df.select_dtypes(include=['number']).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

This heatmap displays the correlation coefficients between all numerical columns in your dataset. Values close to 1 indicate a strong positive correlation, values close to -1 indicate a strong negative correlation, and values close to 0 indicate a weak or no linear correlation. This can help identify strong relationships between different academic performance metrics or other numerical factors.

This count plot displays the distribution of students across different categories (e.g., OBC, GEN, ST, SC). It provides a quick overview of the demographic composition in terms of social categories within your dataset.
